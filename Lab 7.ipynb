```python
import pandas as pd
import matplotlib.pyplot as plt

# imported pandas to read dataset, matplotlib to create visualisations
```

```python
df = pd.read_csv ("malazan_commander_stats (1).csv")

#csv file is loaded into 'df' dataframe, to view data
```

```python
df.head()

#Allows to see first few rows of data
```

```python
#Bar Chart

plt.bar(df["Commander"], df["Battles_Won"],
    color= "plum")


plt.title("Battles Won per Commander")
plt.xlabel("Commander")
plt.ylabel("Battles Won")

plt.xticks(rotation=45)

#plotted a bar chart, compared battles won by commanders
```

```python
#Histogram
plt.hist (df["Avg_Casualty_Rate_%"],
         color = "plum")

plt.title("Distribution of Casualty Rates")
plt.xlabel("Casualty Rate (%)")
plt.ylabel("Frequency")

#Find highest casualty rate

highest = df.loc[df["Avg_Casualty_Rate_%"].idxmax()]

print (highest)

#plotted histogram and highest was by Colatine
```

```python
#Scatter Plot by Faction
for faction, group in df.groupby("Faction"):
    plt.scatter(
        group["Campaigns"],
        group["Battles_Won"],
        label=faction
    )

plt.title("Campaigns Vs Battles Won")
plt.xlabel("Campaigns")
plt.ylabel("Battles Won")

plt.legend()


#after grouping the campaigns, scatter plot shows Malazan Army had most battle wins
```

```python
#Box Plot
plt.boxplot(df["Avg_Casualty_Rate_%"])

plt.title("Casualty Rate Distrbution")

#created boxplot to observe spread of data
```

```python
#Sort By Campaigns

sorted_df = df.sort_values(by = "Campaigns")

#sort dataframe into campaigns column
```

```python
#create line chart

plt.plot(
        sorted_df["Commander"],
        sorted_df["Campaigns"])

plt.title("Campaigns per Commander")
plt.xlabel("Commander")
plt.ylabel("Campaigns")

plt.xticks (rotation=45)


         
```
