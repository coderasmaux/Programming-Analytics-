```python
students = [
    {"name" : "Asmau", "age": 24, "grade": 85},
    {"name": "Hannah", "age": 26, "grade": 77},
    {"name": "Jack", "age": 24, "grade": 90}
]
```

```python
total_grade = 0

for student in students:
    total_grade += student [("grade")]

average_grade = total_grade / len (students)

print ("Average grade:", average_grade)
```

```python
print (students)
print (type(students))
```

```python
print ("Students above average:")

for student in students:
    if student ["grade"] > average_grade:
        print (student["name"])
```

```python
print ("Names in uppercase:")

for student in students:
    print(student["name"].upper())
```

```python
student_name = input ("Enter a student name: ")

try:
    found = False

    for student in students:
        if student ["name"].lower () == student_name.lower ():
            print (student["name"], "grade is", student["grade"])
            found = True
            break

    if not found:
        raise ValueError

except ValueError:
 print ("Error: Student not found.")

```
